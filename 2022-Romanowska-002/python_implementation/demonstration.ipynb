# Out of Africa - Conditional isotropic diffusion (Python)
Demonstration of `2022_romanowska_002_v01.py` (NASSA module 2022-Romanowska-002).

```python
import importlib.util
from pathlib import Path
import matplotlib.pyplot as plt

here = Path.cwd()
spec = importlib.util.spec_from_file_location("oofa", here / "2022_romanowska_002_v01.py")
oofa = importlib.util.module_from_spec(spec)
spec.loader.exec_module(oofa)
```

## Inputs
Values of the NetLogo interface: 20 agents, start point (361, 176), initial radius 2, `pop_growth` 0.30. The world (600 x 351) has the size of the input map.

```python
model = oofa.OutOfAfricaModel(
    number_of_turtles=20,
    input_x=361,
    input_y=176,
    max_initial_distance=2,
    pop_growth=0.30,
    world_width=600,
    world_height=351,
    map_scenario="external_image",
    world_image_path=oofa._default_image_path(),
    random_seed=42,
)
model.setup()
```

## Tick 0

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.imshow(model._render_world(), origin="lower", interpolation="nearest")
ax.set_title("tick 0")
plt.show()
```

## Run 200 ticks

```python
model.run(200)
model.summary()
fig, ax = plt.subplots(figsize=(10, 6))
ax.imshow(model._render_world(), origin="lower", interpolation="nearest")
ax.set_title(f"tick {model.tick_counter}, N = {len(model.turtles)}")
plt.show()
```

## Population over time

```python
plt.plot(model.population_history)
plt.xlabel("tick"); plt.ylabel("agents"); plt.show()
```
